```python
import streamlit as st
import time
# import requests  # 나중에 실제 백엔드 서버와 통신할 때 이 줄의 주석을 해제할 거야

# 백엔드 서버(FastAPI) 역할을 대신할 임시 함수
def mock_fastapi_request(user_input, input_type):
  
    time.sleep(2) # 서버 로딩시간 
    
    # 백엔드 응답규격과 똑같은 형태의 가짜 데이터 
    dummy_response = {
      "status": "success",
      "request_id": "req_20240501_01",
      
      "ml_result": {
        "is_malicious": True,
        "confidence_score": 92.5,
        "model_version": "v1.2"
      },
      
      "vt_result": {
        "detection_ratio": "15/90",
        "malicious_count": 15,
        "harmless_count": 75,
        "link_to_vt": "https://virustotal.com/..."
      },
      
      "ai_agent_report": {
        "summary": "해당 URL은 전형적인 피싱 패턴을 보입니다.",
        "risk_level": "High",
        "steps_to_take": [
          "즉시 접속을 종료하세요.",
          "비밀번호를 변경하세요.",
          "보안팀에 신고하세요."
        ]
      },
      
      "grc_data": {
        "report_id": "GRC-102",
        "download_url": "/api/download/report/102"
      }
    }
    return dummy_response

# 프론트엔드 화면 UI 구성 시작
st.set_page_config(layout="wide")
st.title("지능형 보안 사고 분석 시스템")

# 프론트엔드에서 백엔드로 보낼 요청 데이터 준비
analysis_type = st.radio("분석 대상을 선택하세요:", ["url", "email", "file"])
user_content = st.text_area("분석할 내용을 입력하세요:")

# 분석 시작 버튼
if st.button("분석 요청하기"):
    if user_content:
        with st.spinner("서버에서 데이터를 분석 중입니다..."):
            
            # 임시 함수를 호출하여 확정된 JSON 규격의 데이터를 result 변수에 담음
            result = requests.post((analysis_type, user_content))
            
            #  분석 결과 화면 출력 (받아온 JSON 데이터를 화면에 매칭)
            if result["status"] == "success":
                st.success("분석이 완료되었습니다.")
                
                # 상단: 에이전트 종합 요약
                st.subheader("AI 에이전트 종합 분석")
                st.info(result["ai_agent_report"]["summary"])
                
                # 중단: 화면을 두 칸으로 나누어 ML과 VT 결과를 나란히 배치
                col1, col2 = st.columns(2)
                
                with col1:
                    st.markdown("### 자체 ML 모델 분석 결과")
                    # is_malicious가 True이면 악성, False이면 정상으로 글씨 출력
                    malicious_text = "악성 (Malicious)" if result["ml_result"]["is_malicious"] else "정상 (Safe)"
                    st.metric(label="탐지 결과", value=malicious_text)
                    st.metric(label="위험도 (신뢰도 점수)", value=f"{result['ml_result']['confidence_score']}점")
                    
                with col2:
                    st.markdown("### VirusTotal 교차 검증")
                    st.metric(label="악성 탐지 비율", value=result["vt_result"]["detection_ratio"])
                    st.write(f"악성 판정 엔진 수: {result['vt_result']['malicious_count']}개")
                    st.write(f"정상 판정 엔진 수: {result['vt_result']['harmless_count']}개")
                
                st.markdown("---")
                
                # 하단: 대응 가이드라인 및 리포트 다운로드
                st.subheader("권장 대응 가이드라인")
                
                # JSON 배열(리스트)로 받은 대응 절차를 반복문으로 하나씩 꺼내서 출력
                for index, step in enumerate(result["ai_agent_report"]["steps_to_take"]):
                    st.write(f"{index + 1}. {step}")
                
                st.write("") # 빈 줄 추가
                
                # GRC 파트 다운로드 버튼 구현
                st.download_button(
                    label="기술 리포트 PDF 다운로드",
                    data="임시 리포트 데이터입니다.",
                    file_name=f"{result['grc_data']['report_id']}_report.txt"
                )
    else:
        st.warning("분석할 데이터를 먼저 입력해 주세요.")
```
